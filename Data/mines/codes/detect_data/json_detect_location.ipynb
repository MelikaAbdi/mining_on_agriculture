```python
# specify investment of mine
import re
df['investment'] = str()
for path in os.listdir(dir_path):
    name = int(path.split('.')[0])
    if os.path.isfile(os.path.join(dir_path, path)):
        f = open(dir_path + path,encoding='utf-8')
        data = json.load(f)
        f.close()
        i = data['responses'][0]['textAnnotations']
        inv = str()
        x0 = int()
        y0 = int()
        x = int()
        y = int()
        for n in range(0,len(i)-1):
            if i[n]['description']=='سرمایه' and i[n+1]['description']=='گذاری':
                x0 = i[n+1]['boundingPoly']['vertices'][0]['x']
                y0 = (i[n]['boundingPoly']['vertices'][0]['y'] + i[n]['boundingPoly']['vertices'][2]['y'])/2
                continue
            
            x = i[n]['boundingPoly']['vertices'][0]['x']
            y = (i[n]['boundingPoly']['vertices'][0]['y'] + i[n]['boundingPoly']['vertices'][2]['y'])/2
            if abs(y-y0)<20 and 0<(x0-x)<300:
                inv = i[n]['description'] + inv
                
            if i[n]['description']=='ریال':
                break
                   
        if len(re.findall('\d+',inv))>0:
            df.loc[df['code']==name, 'investment']=unidecode(re.findall('\d+',inv)[0])
            print(df[df['code']==name]['investment'], name, y-y0, x-x0)
```
