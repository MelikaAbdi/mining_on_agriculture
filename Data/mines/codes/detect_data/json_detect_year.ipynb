```python
import pandas as pd
from unidecode import unidecode
import os
import json
import numpy as np

df = pd.DataFrame(columns = ['code', 'license_strt_yr'])

dir_path = r'json files_amar99 and cadastre/json files_amar99 and cadastre/'
c = 0
k = 0

for path in os.listdir(dir_path):
    name = path.split('.')[0]
    ymin = 0
    ymax = 0
    if os.path.isfile(os.path.join(dir_path, path)):
        date = 0
        f = open(dir_path + path,encoding='utf-8')
        data = json.load(f)
        f.close()
        i = data['responses'][0]['textAnnotations']
        for n in range(0,len(i)-1):
            case = i[n]['description']
            if i[n]['description']=='تاریخ' and i[n+1]['description']=='پروانه':
                print(name,i[n]['boundingPoly']['vertices'][0],i[n]['boundingPoly']['vertices'][2])
                ymin = i[n]['boundingPoly']['vertices'][0]['y']
                ymax = i[n]['boundingPoly']['vertices'][2]['y']
            tre = (ymax - ymin)*1.5
            if abs(i[n]['boundingPoly']['vertices'][0]['y']-ymin)<tre and abs(i[n]['boundingPoly']['vertices'][2]['y']-ymax)<tre and case!=':' and case!='پروانه' and case[0]=='۱' and (len(case)==4 or ('/' in case and len(case)==10)):
                date = case.split('/')[0]
                date = unidecode(date)
                print(date)

        list_row = [int(name), int(date)]
        df.loc[len(df)] = list_row
        k = k+1
        if date==str():
            c = c+1
            
```
