```python
#finding coordinates of desired data
import re
import pandas as pd
import os
import json
from difflib import SequenceMatcher
coordinates = pd.DataFrame(columns = ['code','mannual','x_isic','y_isic','x_storage','y_storage','x_pstorage', 'y_pstorage', 'x_extraction', 'y_extraction', 'x_cu1', 'y_cu1', 'x_cu2', 'y_cu2'])
dir_path_ore = "final list of mines/licenses_ore_json/"
for path in os.listdir(dir_path_ore):
    name = path.split('.')[0]
    if os.path.isfile(os.path.join(dir_path_ore, path)):
        f = open(dir_path_ore + path,encoding='utf-8')
        data = json.load(f)
        f.close()
        i = data['responses'][0]['textAnnotations']
        
        x_isic = int()
        x_s = int()
        x_ps = int()
        x_ex = int()
        x_cu1 = int()
        x_cu2 = int()
        y_isic = int()
        y_s = int()
        y_ps = int()
        y_ex = int()
        y_cu1 = int()
        y_cu2 = int()
        s = 0
        ps = 0
        ex = 0
        isic = 0
        cu_1 = 0
        cu_2 = 0
        m = 0
        bool1 = False
        bool2 = False
        bool3 = False
        f1 = False
        f2 = False
        f3 = False
        
        for n in range(0,len(i)-2):
            bool1 = (SequenceMatcher(None,'ذخیرهقطعی', i[n]['description'] + i[n+1]['description']).ratio()>=0.8)
            bool2 = (SequenceMatcher(None,'ذخیرهاحتمالی', i[n]['description'] + i[n+1]['description']).ratio()>=0.8)
            bool3 = (SequenceMatcher(None,'استخراجسالیانه', i[n]['description'] + i[n+1]['description']).ratio()>=0.8)
            
            if (bool1) & (f1 & f2 & f3):
                m = 1
                coordinates.loc[len(coordinates)] = [name,m,x_isic,y_isic,x_s,y_s,x_ps,y_ps,x_ex,y_ex,x_cu1,y_cu1,x_cu2,y_cu2]
                
            cu1 = SequenceMatcher(None,'عیارحد', i[n]['description'] + i[n+1]['description']).ratio()
            cu2 = SequenceMatcher(None,'عیارمیانگین', i[n]['description'] + i[n+1]['description']).ratio()
            isic = SequenceMatcher(None,'کدآیسیک', i[n]['description'] + i[n+1]['description']).ratio()
            s = SequenceMatcher(None,'ذخیرهقطعی', i[n]['description'] + i[n+1]['description']).ratio()
            ps = SequenceMatcher(None,'ذخیرهاحتمالی', i[n]['description'] + i[n+1]['description']).ratio()
            ex = SequenceMatcher(None,'استخراجسالیانه', i[n]['description'] + i[n+1]['description']).ratio()
            
            if cu1>=0.8:
                x_cu1 = (i[n+1]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
                y_cu1 = i[n+1]['boundingPoly']['vertices'][0]['y']
                
            if cu2>=0.8:
                x_cu2 = (i[n+1]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
                y_cu2 = i[n+1]['boundingPoly']['vertices'][0]['y']
            
            if isic>=0.8:
                x_isic = (i[n+1]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
                y_isic = i[n+1]['boundingPoly']['vertices'][0]['y']
                
            if s>=0.8:
                x_s = (i[n+1]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
                y_s = i[n+1]['boundingPoly']['vertices'][0]['y']
                f1 = True
                
            if ps>=0.8:
                x_ps = (i[n+1]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
                y_ps = i[n+1]['boundingPoly']['vertices'][0]['y']
                f2 = True
                
            if ex>=0.8:
                x_ex = (i[n+1]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
                y_ex = i[n+1]['boundingPoly']['vertices'][0]['y']
                f3 = True
                
        coordinates.loc[len(coordinates)] = [name,m,x_isic,y_isic,x_s,y_s,x_ps,y_ps,x_ex,y_ex,x_cu1,y_cu1,x_cu2,y_cu2]
        print([name,m,x_isic,x_s,x_ps,x_ex])
        
```

```python
#gathering desired data
import re
from unidecode import unidecode
from difflib import SequenceMatcher
#df_ore = pd.DataFrame(columns = ['code','isic', 'storage', 'pro_storage', 'extraction', 'limit_cutie', 'ave_cutie'])
df_ore['isic'] = str()
df_ore['storage'] = str()
df_ore['pro_storage'] = str()
df_ore['extraction'] = str()
dir_path_ore = "final list of mines/licenses_ore_json/"
for path in coordinates['code']:
    name = path.split('.')[0]
    if True:
        f = open(dir_path_ore + path+'.json',encoding='utf-8')
        data = json.load(f)
        f.close()
        i = data['responses'][0]['textAnnotations']
        
        isic = str()
        s = str()
        ps = str()
        ex = str()
        cu1 = str()
        cu2 = str()
        gap_s = int()
        gap_ps = int()
        gap_ex = int()
        x_gap_isic = 1000
        x_gap_s = 1000
        x_gap_ps = 1000
        x_gap_ex = 1000
        x_gap_cu1 = 1000
        x_gap_cu2 = 1000
        y_gap_isic = 1000
        y_gap_s = 1000
        y_gap_ps = 1000
        y_gap_ex = 1000
        y_gap_cu1 = 1000
        y_gap_cu2 = 1000
        
        x_isic = coordinates[coordinates['code']==name]['x_isic'].reset_index(drop=True)
        x_s = coordinates[coordinates['code']==name]['x_storage'].reset_index(drop=True)
        x_ps = coordinates[coordinates['code']==name]['x_pstorage'].reset_index(drop=True)
        x_ex = coordinates[coordinates['code']==name]['x_extraction'].reset_index(drop=True)
        x_cu1 = coordinates[coordinates['code']==name]['x_cu1'].reset_index(drop=True)
        x_cu2 = coordinates[coordinates['code']==name]['x_cu2'].reset_index(drop=True)
        y_isic = coordinates[coordinates['code']==name]['y_isic'].reset_index(drop=True)
        y_s = coordinates[coordinates['code']==name]['y_storage'].reset_index(drop=True)
        y_ps = coordinates[coordinates['code']==name]['y_pstorage'].reset_index(drop=True)
        y_ex = coordinates[coordinates['code']==name]['y_extraction'].reset_index(drop=True)
        y_cu1 = coordinates[coordinates['code']==name]['y_cu1'].reset_index(drop=True)
        y_cu2 = coordinates[coordinates['code']==name]['y_cu2'].reset_index(drop=True)
        
        for n in range(0,len(i)):
            
            if (s!= str() and abs(y_s[0]-i[n]['boundingPoly']['vertices'][0]['y'] - y_gap_s)>5) or ('|' in i[n]['description']) or ('ز' in i[n]['description']) or ('م' in i[n]['description']) or ('/' in i[n]['description']) or ('ح' in i[n]['description']) or ('ن' in i[n]['description']) or ('ل' in i[n]['description']) or ('ر' in i[n]['description']) or ('ت' in i[n]['description']) or ('ی' in i[n]['description']) or ('ه' in i[n]['description']) or (')' in i[n]['description']) or ('(' in i[n]['description']):  
                continue
            
            x_center = (i[n]['boundingPoly']['vertices'][0]['x'] + i[n]['boundingPoly']['vertices'][1]['x'])/2
            
            if len(x_isic)>0:
                x_gap_isic = abs(x_isic[0]-x_center)
                y_gap_isic = (y_isic[0]-i[n]['boundingPoly']['vertices'][0]['y'])
            if len(x_s)>0:
                x_gap_s = abs(x_s[0]-x_center)
                y_gap_s = (y_s[0]-i[n]['boundingPoly']['vertices'][0]['y'])
            if len(x_ps)>0:
                x_gap_ps = abs(x_ps[0]-x_center)
                y_gap_ps = (y_ps[0]-i[n]['boundingPoly']['vertices'][0]['y'])
            if len(x_ex)>0:
                x_gap_ex = abs(x_ex[0]-x_center)
                y_gap_ex = (y_ex[0]-i[n]['boundingPoly']['vertices'][0]['y'])
            if len(x_cu1)>0:
                x_gap_cu1 = abs(x_cu1[0]-x_center)
                y_gap_cu1 = (y_cu1[0]-i[n]['boundingPoly']['vertices'][0]['y'])
            if len(x_cu2)>0:
                x_gap_cu2 = abs(x_cu2[0]-x_center)
                y_gap_cu2 = (y_cu2[0]-i[n]['boundingPoly']['vertices'][0]['y'])
            
            if x_gap_isic<30 and x_s[0]!=0 and y_gap_isic<-20:
                isic = isic+i[n]['description']
            if x_gap_s<30 and x_s[0]!=0 and y_gap_s<-20:
                s = s+i[n]['description']
            if x_gap_ps<30 and x_ps[0]!=0 and y_gap_ps<-20:
                ps = ps+i[n]['description']
            if x_gap_ex<30 and x_ex[0]!=0 and y_gap_ex<-20:
                ex = ex+i[n]['description']
            if x_gap_cu1<30 and x_cu1[0]!=0 and y_gap_cu1<-20:
                cu1 = cu1+i[n]['description']
            if x_gap_cu2<30 and x_cu2[0]!=0 and y_gap_cu2<-20:
                cu2 = cu2+i[n]['description']
                
        
            s = s.replace('*','۰')
            s = s.replace('F','۴')
            s = s.replace('0','۵')
            s = s.replace('Y','۳')
            s = s.replace('.','۰')
            s = s.replace('+','۰')
            s = s.replace('9','۹')
            s = s.replace('1','۱')
            s = s.replace('A','۸')
            s = s.replace('T','۲')

            ps = ps.replace('*','۰')
            ps = ps.replace('F','۴')
            ps = ps.replace('0','۵')
            ps = ps.replace('Y','۳')
            ps = ps.replace('.','۰')
            ps = ps.replace('+','۰')
            ps = ps.replace('9','۹') 
            ps = ps.replace('1','۱')
            ps = ps.replace('A','۸')
            ps = ps.replace('T','۲')

            ex = ex.replace('*','۰')
            ex = ex.replace('F','۴')
            ex = ex.replace('0','۵')
            ex = ex.replace('Y','۳')
            ex = ex.replace('.','۰')
            ex = ex.replace('+','۰')
            ex = ex.replace('9','۹') 
            ex = ex.replace('1','۱')
            ex = ex.replace('A','۸')
            ex = ex.replace('T','۲')
            
            
            #s = re.findall(r'[\d.+*]+|[.+*]',s)[0]
            #ps = re.findall(r'[\d.+*]+|[.+*]',ps)[0]
            #ex = re.findall(r'[\d.+*]+|[.+*]',ex)[0]
        df_ore.loc[df_ore['code']==int(re.findall(r'[\d.+*]+|[.+*]',name)[0]),'isic'] = unidecode(isic)
        df_ore.loc[df_ore['code']==int(re.findall(r'[\d.+*]+|[.+*]',name)[0]),'storage'] = unidecode(s)
        df_ore.loc[df_ore['code']==int(re.findall(r'[\d.+*]+|[.+*]',name)[0]),'pro_storage'] = unidecode(ps)
        df_ore.loc[df_ore['code']==int(re.findall(r'[\d.+*]+|[.+*]',name)[0]),'extraction'] = unidecode(ex)
        df_ore.loc[df_ore['code']==int(re.findall(r'[\d.+*]+|[.+*]',name)[0]),'limit_cutie'] = unidecode(cu1)
        df_ore.loc[df_ore['code']==int(re.findall(r'[\d.+*]+|[.+*]',name)[0]),'ave_cutie'] = unidecode(cu2)

        
        #df_ore.loc[len(df_ore)] = [re.findall(r'[\d.+*]+|[.+*]',name)[0],unidecode(isic),unidecode(s),unidecode(ps),unidecode(ex)]
        print([re.findall(r'[\d.+*]+|[.+*]',name)[0],unidecode(isic),unidecode(s),unidecode(ps),unidecode(ex)])
        
```
