```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time 
from selenium.webdriver.common.keys import Keys
import numpy as np
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains

df = pd.DataFrame(pd.read_csv('ime_mines.csv'))
df['cad_state']= np.ones(len(df))

driver = webdriver.Chrome("chromedriver")
driver.get("https://cadastre.mimt.gov.ir/QR.aspx")

for cc in range(789,len(df)):
    typetextfirst = driver.find_element("xpath",'//input[@id="ContentPlaceHolder1_txtMineno"]')
    typetextfirst.clear()
    typetextfirst.send_keys(str(df.code[cc]))
    botton = driver.find_element("xpath",'//input[@id="ContentPlaceHolder1_btnSearch"]')
    botton.click()
    time.sleep(1)
    df['cad_state']=1
    if driver.find_elements("xpath",'//span[@class="rwInnerSpan"]'):
        botton = driver.find_element("xpath",'//span[@class="rwInnerSpan"]')
        if botton.is_displayed():
            botton.click()
            df['cad_state'][cc]=0
```
